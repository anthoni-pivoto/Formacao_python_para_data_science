```python
import pandas as pd
```

```python
dados = pd.read_json('ex_dados_vendas_clientes.json')
dados.head()
```

```python
dados = pd.json_normalize(dados['dados_vendas'])
```

```python
dados.head()
```

```python
colunas = list(dados.columns)
```

```python
dados = dados.explode(colunas[1:])
dados.reset_index(drop=True, inplace=True)
```

```python
dados
```

```python
dados['Valor da compra'] = dados['Valor da compra'].apply(lambda x: x.replace('R$','').replace(',','.').strip())
```

```python
import numpy as np
```

```python
dados['Valor da compra'] = dados['Valor da compra'].astype(np.float64)
```

```python
dados['Cliente'] = dados['Cliente'].str.replace('[^a-zA-Z ]','',regex=True).str.strip().str.upper()
dados
```
